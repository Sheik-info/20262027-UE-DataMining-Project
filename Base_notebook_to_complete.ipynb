```python
import pandas as pd
pd.set_option('display.max_columns', None)

pa = pd.read_csv('data/health/patient_administration.csv')
pa.head()
```

```python
# voir la ligne dupliquée
print(pa[pa.duplicated(keep=False)])

# la supprimer
pa = pa.drop_duplicates().reset_index(drop=True)
print(pa.shape)  # doit passer de 241 à 240 lignes
```

```python
for col in ['registered_sex', 'record_status', 'home_region']:
    print(col, '->', pa[col].unique())
```

```python
# record_status : uniformisation en minuscules ('ACTIVE' -> 'active')
pa['record_status'] = pa['record_status'].str.lower()

# home_region : uniformisation en casse "Titre" ('vALENNE rEPUBLIC' -> 'Valenne Republic')
pa['home_region'] = pa['home_region'].str.title()

# vérification après correction : chaque valeur ne doit plus apparaître qu'une fois
print(pa['record_status'].unique())
print(pa['home_region'].unique())
```

```python
# Conversion de la date de naissance et calcul de l'âge actuel
pa['date_of_birth'] = pd.to_datetime(pa['date_of_birth'])
today = pd.Timestamp.now()
pa['age'] = (today - pa['date_of_birth']).dt.days // 365

# On vérifie qu'il n'y a pas d'âge aberrant (négatif ou > 120 ans par exemple)
print(pa['age'].describe())
```

```python
# Diagnostic : on affiche les valeurs uniques de registration_date pour repérer les formats mélangés
print(pa['registration_date'].unique())
```

```python
# On isole les lignes avec un format de date non-standard pour les inspecter
mask_anomalie = ~pa['registration_date'].str.match(r'^\d{4}-\d{2}-\d{2}$')
print(pa.loc[mask_anomalie, 'registration_date'])
```

```python
# Correction manuelle des 2 formats détectés (YYYY/MM/DD et DD/MM/YYYY)
# En s'appuyant sur le fait que toutes les autres dates de la colonne ont pour jour "01",
# ces deux valeurs représentent la même date : le 1er septembre 2019
pa['registration_date'] = pa['registration_date'].replace({
    '2019/09/01': '2019-09-01',
    '01/09/2019': '2019-09-01'
})

# Conversion en datetime maintenant que le format est uniforme
pa['registration_date'] = pd.to_datetime(pa['registration_date'], format='%Y-%m-%d')

# Vérification de cohérence : la date d'inscription doit être postérieure à la naissance
incoherent = pa[pa['registration_date'] < pa['date_of_birth']]
print(incoherent)
```
